# Einheit 02: Numerische Verfahren zum Lösen von ODEs (=gewöhnliche Differentialgleichungen)
In dieser Einheit geht es darum, Differentialgleichungen mit Hilfe numerischer Verfahren zu lösen. Behandelt werden das aus der Vorlesung bekannte Euler-Verfahren sowie das ähnliche Heun-Verfahren.

## Numerische Verfahren für ODEs der Ordnung 1
Zuerst behandeln wir exakt lösbare Differentialgleichungen und vergleichen exakte Lösungen mit expliziten Euler-Lösungen.

### Präsenzaufgaben
Wir wissen bereits, dass lineare ODEs 1. Ordnung $y' = f(t) y + g(t)$, $t \in [t_0,t_1]$, Lösungen der Form $\mathrm{e}^{F(t)} \cdot (c + G(t))$ besitzen, wobei $F$ eine Stammfunktion von $f$ und $G$ eine Stammfunktion von $\mathrm{e}^{-F} \cdot g$ ist und $c \in {\mathbb R}$ eine Konstante.

1. Seien $t_0 = 0, t_1 = 2 \pi.$ Bestimmen Sie für das AWP (=Anfangswertproblem) 
    $$y' = \cos(t) \cdot y, \quad y(0) = 1$$ 
    die exakte Lösung und erstellen Sie einen Plot der Lösung.
2. (vergleiche VL) Implementieren Sie das Heunverfahren zur approximativen Lösung des AWPs zur ODE 1. Ordnung
    $$ y' = f(y,t), ~ y(0) = y_0, t \in [a,b] $$
    Definieren Sie dazu eine Funktion `Heun(f,t,y0)`, wobei `t` die Liste der Stellen ist, `f` $= f(y,t)$ die geg. Funktion und `y0` der Startwert $y(0)$ des AWPs sind.    
3. Erweitern Sie Ihren Plot aus 1., indem Sie für $h \in \{ 1, 1/2, 1/4, 1/10 \}$ und `t= np.arrange(0,2*np.pi,h)` die Heun-Lösung `Heun(f,t,y0)` in das Schaubild einzeichnen. Was können Sie über die maximalen Abweichungen der Heun-Lösungen sagen?

## Numerische Verfahren für harmonische DGL-Systeme
In der Vorlesung haben wir gesehen, dass $(x(t),y(t))$ genau dann das Hamiltonsche System 
    $$ x' = \frac{\partial H}{\partial y}, \quad y' = - \frac{\partial H}{\partial x} $$
löst, wenn $H(x(t),y(t))$ konstant in $t$ ist.

### Präsenzaufgaben
4. Implementieren Sie das expliziten Eulerverfahren für zu DGL-Systemen 1. Ordnung gehörende AWPs der Form
    $$ x' = f(x,y,t), ~ y' = g(x,y,t), ~ x(0) = x_0, ~ y(0) = y_0 $$
    Definieren Sie dazu eine Funktion `exEuler(f,g,t,x0,y0)`, wobei `t` die Liste der Stellen ist, `f` $= f(x,y,t)$ und `g` $= g(x,y,t)$ die geg. Funktionen und `x0,y0` die geg. Startwerte des AWPs sind.

Betrachten Sie das Hamiltonsche System
    $$ x' = y, \quad y' = - x, ~ x(0) = 1, ~ y(0) = 0  $$
mit zugehöriger Hamilton-Funktion $H(x,y) = \frac{x^2}{2} + \frac{y^2}{2}$.

5. Erstellen Sie für dieses System für $h = 1/10$ und `t=np.arrange(0,100,h)` einen Plot, der beide Komponenten `x_Euler` und `y_Euler` von `exEuler(f,g,t,1,0)` in einem gemeinsamen Schaubild gegen `t` aufträgt.     
6. Erstellen Sie für das selbe `t` ein Phasenportrait für die explizite Euler-Lösung `exEuler(f,g,t,1,0)`, d.h. ein Schaubild, indem die `y_Euler`-Werte gegen die `x_Euler`-Werte aufgetragen werden.

### Hausaufgaben
4. Implementieren Sie das Heunverfahren für zu DGL-Systemen 1. Ordnung gehörende AWPs der Form
    $$ x' = f(y,t), ~ y' = g(x,t), ~ x(0) = x_0, ~ y(0) = y_0. $$
    Definieren Sie dazu eine Funktion `Heun(f,g,t,x0,y0)`, wobei `t` die Liste der Stellen ist, `f` $= f(y,t)$ und `g` $= g(x,t)$ die geg. Funktionen und `x0,y0` die geg. Startwerte des AWPs sind.

Betrachten Sie das logarithmisierte Lotka-Volterra-Räuber-Beute-System aus der Vorlesung
    $$ x' = \alpha - \beta \cdot \mathrm{e}^{y(t)}, ~ y' = - \gamma + \delta \cdot \mathrm{e}^{x(t)}, ~ x(0) = x0 = \ln(50000), ~ y(0) = y0 = \ln(500) $$
    mit Parametern 
    $\alpha = 2,~
    \beta = 2\cdot 10^{-3},~
    \gamma = 0.2,~
    \delta = 2\cdot10^{-6}.$
    
5. Erstellen Sie für dieses System für $h = 1/10$ und `t=np.arrange(0,100,h)` einen Plot, der die beiden Komponenten `x_Heun` und `y_Heun` von `(f,g,t,1,0)` in einem gemeinsamen Schaubild gegen `t` aufträgt. 
6. Erstellen Sie für das selbe `t` ein Phasenportrait für die Heun-Lösung `Heun(f,g,t,1,0)`, d.h. ein Schaubild, indem die `y_Heun`-Werte gegen die `x_Heun`-Werte aufgetragen werden.
