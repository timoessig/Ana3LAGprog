```python
# Antwortfeld zu Präsenzaufgabe 2
def Heun(f,t,y0):
    y = [y0]
    for n in range(0,len(t)-1):
        z_n = y[n]+ (t[n+1]-t[n])*f(t[n],y[n])
        y_new = y[n] + (t[n+1]-t[n])/2 * (f(t[n],y[n]) + f(t[n+1],z_n))
        y.append(y_new)
    return t,y
```
