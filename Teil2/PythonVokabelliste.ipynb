```python
# Python-Tutorial: Arbeiten mit Listen und Plotten mit Matplotlib

# --- TEIL 1: Arbeiten mit Listen ---

# 1. Eine Liste erstellen
meine_liste = [1, 2, 3, 4, 5]

# 2. Elemente per Index (0-basierte Indizierung) abrufen
erstes_element = meine_liste[0]
letztes_element = meine_liste[-1]

# 3. Ein Element zu einer Liste hinzufügen
meine_liste.append(6)  # [1, 2, 3, 4, 5, 6]

# 4. Ein Element per Wert entfernen
meine_liste.remove(3)  # [1, 2, 4, 5, 6]

# 5. Eine Liste teilen (Subliste erstellen)
teil_liste = meine_liste[1:4]  # [2, 4, 5]

# 6. Durch eine Liste iterieren
for element in meine_liste:
    print(element)

# 7. List Comprehensions (eine neue Liste erstellen)
quadrate_liste = [x**2 for x in meine_liste]  # [1, 4, 16, 25, 36]

# 8. Eine Liste mit konstanten Werten erstellen
x0 = 5
t = [0, 1, 2, 3, 4]
x0liste = [x0] * len(t)  # [5, 5, 5, 5, 5]

# --- TEIL 2: Plotten mit Matplotlib ---

import numpy as np
import matplotlib.pyplot as plt

# 1. Daten für das Plotten erstellen
t = np.linspace(0, 10, 100)  # 100 Punkte zwischen 0 und 10 generieren
y = np.sin(t)  # Sinusfunktion als Beispieldaten

# 2. Einfaches Liniendiagramm
plt.plot(t, y, label='sin(t)')
plt.title("Einfaches Liniendiagramm")
plt.xlabel("Zeit (t)")
plt.ylabel("Amplitude")
plt.legend()
plt.grid()
plt.show()

# 3. Mehrere Plots in einer Grafik
z = np.cos(t)  # Kosinusfunktion
plt.plot(t, y, label='sin(t)')
plt.plot(t, z, label='cos(t)', linestyle='--')
plt.title("Mehrere Linien")
plt.xlabel("Zeit (t)")
plt.ylabel("Amplitude")
plt.legend()
plt.grid()
plt.show()

# 4. Streudiagramm
plt.scatter(t, y, label='sin(t)', color='red')
plt.title("Streudiagramm")
plt.xlabel("Zeit (t)")
plt.ylabel("Amplitude")
plt.legend()
plt.grid()
plt.show()

# 5. Anpassung des Diagramms
plt.plot(t, y, label='sin(t)', linewidth=2, color='purple')
plt.title("Angepasstes Diagramm")
plt.xlabel("Zeit (t)")
plt.ylabel("Amplitude")
plt.legend()
plt.grid(alpha=0.3)
plt.show()

# --- TEIL 3: Wichtige Konstanten und Funktionen in Numpy ---

# Wichtige Konstanten
pi = np.pi  # Kreiszahl π
E = np.e  # Euler'sche Zahl e
unendlich = np.inf  # Unendlichkeit

# Wichtige mathematische Funktionen
wurzel = np.sqrt(16)  # Quadratwurzel: 4.0
sinus = np.sin(pi / 2)  # Sinus: 1.0
kosinus = np.cos(0)  # Kosinus: 1.0
exponential = np.exp(1)  # Exponentialfunktion: e^1
logarithmus = np.log(10)  # Natürlicher Logarithmus: ln(10)
logarithmus10 = np.log10(100)  # Logarithmus Basis 10: log10(100)

# Funktionen für Arrays
array = np.array([1, 2, 3])  # Array erstellen
zeros = np.zeros(5)  # Array mit Nullen: [0. 0. 0. 0. 0.]
ones = np.ones(3)  # Array mit Einsen: [1. 1. 1.]
linspace = np.linspace(0, 1, 5)  # Gleichmäßig verteilte Werte: [0.   0.25 0.5  0.75 1.  ]
arange = np.arange(0, 10, 2)  # Werte mit festem Schritt: [0 2 4 6 8]
random = np.random.random(3)  # Zufällige Werte zwischen 0 und 1

# Statistische Funktionen
mittelwert = np.mean(array)  # Mittelwert: 2.0
summe = np.sum(array)  # Summe: 6
minimum = np.min(array)  # Minimum: 1
maximum = np.max(array)  # Maximum: 3
```
