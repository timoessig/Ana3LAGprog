### 1. Falsche Abhängigkeiten bei Funktionen
Hängen Funktionen von bestimmten Größen ab, so sollten alle diese Größen betreffenden Hilfsgrößen innerhalb der Funktion definiert werden.

**Fehler 1: x-Liste außerhalb der Definition festgelegt**
Hier wollen wir eine Funktion `sum(a,b)` definieren, die für ganze Zahlen $a < b$ alle Zahlen von $a$ bis $b$ summiert.

```python
a = 1
b = 5 
x = range(a,b+1)

def sum(a,b):
    return np.sum(x)

# Test der Funktion:
sum(1,2)
```

Man erhält hier zwar ein Ergebnis, aber es ist falsch. `sum(1,2)` sollte $1+2=3$ ausgeben.
Aber warum erhalten wir einen falschen Wert?

Der Fehler liegt darin, dass `x = range(a,b+1)` (die Menge aller ganzen Zahlen von $a$ bis $b$ vor der eigentlich Definition festgesetzt wird. Damit wird für die bisher im Zwischenspeicher liegenden Werte von a,b (hier: a = 1, b = 5) die entsprechende Liste erstellt, danach aber nicht mehr geändert. Richtig wäre:

```python
def sum(a,b):
    x = range(a,b+1)
    return np.sum(x)

# Test der Funktion:
sum(1,2)
```

Der Wert überzeugt!

**Fehler 2:**

```python
a = 1
b = 5
def Summe_falscheAbh(b):
    x = range(a,b+1)
    return np.sum(x)

# Test der Funktion:
Summe_falscheAbh(2)
```

Hier ist zwar das Ergebnis richtig, doch die Abhängigkeiten sind nicht korrekt benannt. Die Funktion `Summe_falscheAbh` hängt laut Bezeichnung nur von `b` ab, in Wahrheit aber ebenso von `a`. 
Diese Syntax ist problematisch, weil sie sehr fehleranfällig ist. Wir müssen so vor dem Aufruf der Funktion immer `a` angeben, damit das Ergebnis richtig ist.

### 1. Einrückungen bei Funktionen und Schleifen
Bei `if`-Abfragen, Schleifen und der Definition von Funktionen sind Einrückungen essentiell. Generell gilt: Es wird um eine Tabulatorlänge bzw. 4 Leerzeichen eingerückt. Wird nicht korrekt eingerückt, gibt der Code entweder eine Fehlermeldung aus oder, was meist schlimmer ist, funktioniert nicht, wie man möchte. Nachfolgend einige Beispiele.

i) *`if`-Schleife mit Fehlermeldung weil falsch eingerückt*

```python
x = 10
if x > 5:
print("x is greater than 5")
```

Hier fehlt die Einrückung um 4 Leerzeichen in der dritten Zeile.

ii) *`if`-Schleife, falsch eingerückt, ohne Fehlermeldung*

```python
x = 10
if x > 5:
    y = x +1
    print(f"x+1 ist {y}")
print("x ist größer als 5")
```

Die Ausgabe scheint korrekt, doch wenn `x` die Ungleichung nicht erfüllt, passiert folgendes:

```python
x = 1
if x > 5:
    y = x +1
    print(f"x+1 ist {y}")
print("x ist größer als 5")
```

Obwohl $x<5$ ist, wird der Antwortsatz "x ist größer als 5" fälschlicherweise ausgegeben. Der Grund: Er ist nicht eingerückt. 
Es gibt keine Fehlermeldung, weil nach dem Aufruf der `if`-Klausel eine korrekt eingerückte Zeile steht.

iii) *for*-Schleife

```python
Früchte = ["Apfel", "Birne", "Heidelbeere"]
for Frucht in Früchte:
    print('Hier die Liste der Früchte:')
print(Frucht)
```

Das tut ja überhaupt nicht was wir wollen und wir erhalten nicht einmal eine Fehlermeldung. Was tut diese Schleife:
Sie geht der Reihe nach die drei Früchte durch und gibt für jede Frucht den Satz `Hier die Liste der Früchte:` aus. Da die letzte Frucht `Heidelbeere` ist, bleibt diese im Zwischenspeicher als `Frucht` gespeichert und wird deshalb am Ende, nach Durchlaufen der Schleife noch einmal ausgegeben.

Sinnvoller wäre:


```python
Früchte = ["Apfel", "Birne", "Heidelbeere"]
print('Hier die Liste der Früchte:')
for Frucht in Früchte:
    print(Frucht)
```

iv) *Funktionen mittels `def`* 

Wie bei den Schleifen zuvor, folgt nach der `def ...:` Zeile eine Einrückung. Beispiel:

```python
def FlächeRechteck(Breite, Höhe):
    Fläche = Breite * Höhe
    return Fläche

# Test:
FlächeRechteck(2,5)
```

v) *Verschachtelte Schleifen und Definitionen* 

Hat man Verschachtelungen von Schleifen und Definitionen, erhält man verschiedene Einrückungen. 
Generell gilt: Zu jedem `def`,`if`, `for`,... Block gehört eine Einrückung für die Inhalte dieses Blocks.

Beispiel: Wir definieren eine Funktion, die aus einer Liste von Zahlen alle geraden Zahlen herausfiltert und diese in Zahlen $<4$ und $\geq 4$ unterteilt.

```python
#L soll eine Liste sein
def gerader4filter(L):
    for item in L:
        if item % 2 == 0:
            if item < 4:
                print(f"{item} ist gerade und kleiner als 4")
            else:
                print(f"{item} ist gerade und größer gleich 4")
        else:
            print(f"{item} ist keine gerade ganze Zahl")
        
L = [1, 2, 3, 4, 5, 6, 0.1]
gerader4filter(L)
```
