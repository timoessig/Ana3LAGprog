### 1. Verschiedene Arten von Schleifen

Wir wollen an folgendem Beispiel zeigen, wie man eine Iteration auf verschiedene Arten implementieren kann.

**Ziel:** Wurzel 2 numerisch approximieren

**Vorgehen:**
1. Gegeben ist der Startwert s = 1
2. Im n-ten Schritt wird geschaut, ob $s^2$ größer oder kleiner als $\sqrt{2}$ ist. Falls kleiner, addieren wir $1/n$, falls größer, subtrahieren wir $1/n$, falls gleich, hören wir auf.
3. Das ganze machen wir 1000 Mal

```python
# Methode 1: for-Schleife
def annaeherung_for(start):
    # Wir starten mit s=startwert
    s = start
    # der Reihe nach führen wir für n < 1001 etwas durch
    for n in range(1,1001):
        if s**2 == 2:
            return s
        elif s**2 < 2:
            # in diesem Fall wird 1/n zu s addiert, d.h. s wird ersetzt durch s + 1/n
            s += 1/n
        elif s**2 > 2:
            # in diesem Fall wird 1/n von s subtrahiert, d.h. s wird ersetzt durch s - 1/n
            s += -1/n
    return(s)

annaeherung_for(1)
```

```python
# Methode 2: while-Schleife
def annaeherung_while(start):
    # Wir starten mit s=startwert
    s = start
    n = 1
    # beginnend mit n = 1 führen wir folgende Operationen durch, falls n =< 1000
    while n <= 1000:
        if s**2 == 2:
            return s
        elif s**2 < 2:
            # in diesem Fall wird 1/n zu s addiert, d.h. s wird ersetzt durch s + 1/n
            s += 1/n
        elif s**2 > 2:
            # in diesem Fall wird 1/n von s subtrahiert, d.h. s wird ersetzt durch s - 1/n
            s += -1/n
        # Nach dem Durchlauf wird der n-Wert um 1 erhöht
        n += 1
    return(s)

annaeherung_while(1)
```

```python
# Methode 3: Rekursive Funktion
def annaeherung_rekursiv(start):
    # Man definiert eine rekursive Hilfsfunktion,
    # abhängig vom Iterationsschritt n und der Variable t
    def h(n,t):
        # falls n = 1000 ist oder t^2 = 2 erfüllt, hört man auf und gibt t aus
        if n == 1000 or t**2 == 2:
            return t
        # falls t^2 < 2 ist, führt man die Funktion rekursiv aus, mit n+1 statt n und t+1/n statt t
        elif t**2 < 2:
            return h(n+1,t+1/n)
        # falls t^2 > 2 ist, führt man die Funktion rekursiv aus, mit n+1 statt n und t-1/n statt t
        elif t**2 > 2:
            return h(n+1,t-1/n)
    # Zum Schluss ruft man die rekursive Hilfsfunktion h auf, mit Startwerten n = 1 und t = start
    return h(1,start)

annaeherung_rekursiv(1)
```
